### Part 1: Fonksiyonlar

Fonksiyonlar tıpkı matematikte olduğu gibi girdi ve çıktıya sahip olan ve bizler için çok sihirli işlevler gerçekleştirir. Nasıl $f(x) = x + 5$ gibi ifadeler yazıp matematiksel anlamda önemli işlemleri ifade ediyorsak, Python'da ve diğer kodlama dillerinde de aynı şekilde çeşitli işlemleri fonksiyonla ifade edebiliriz. Peki, kodumuzu normal bir şekilde de yazabilirken neden fonksiyonlara ihtiyaç duyayım ki? Bunun tam olarak **4** adet temel sebebi var:

1. **Tekrar Kullanılabilirlik (*Reusability*):** Bazen aynı işlemi defalarca yapabiliriz. Bir kod içerisinde 10 farklı yerde kullanmamız gereken bir prosedür olduğunu düşünelim. Ben eğer bunu her seferinde tekrar yazarsam zaman kaybına yol açarım. Bunun yanı sıra bir 10 kez daha kullanmam gerektiğinde her seferinde tekrar yazmam gerekir. İşte bundan kaçınmamız lazım.
2. **Modülerlik (*Modularity*):** Bir işlemi her kullanmamız gerektiğinde onu tekrar yazmak yerine bir fonksiyon kullanmak tıpkı lego parçalarını birleştirmeye benzer. Her seferinde aynı kod bloğunu yazmak yerine o kod bloğunu rahatlıkla enjekte etmemize olanak tanır.
3. **Kod Okunurluğu (*Readability*):** Kullandığım kod bloğunun 50 satırdan oluştuğunu düşünelim. Bunu 10 kez kullandığımda 10 x 50 = 500 satırlık bir kod yığını oluşturmuş olurum. İnanın, bunu kimse okumak da istemez, kodu anlaması da zorlaşır. Bunun yerine basitçe fonksiyonu yazıp kullanırsam kodun anlaşılması da o kadar kolaylaşır.
4. **Sürdürülebilirlik (*Maintainability*):** Kodtabanları yıllar boyunca tonlarca insan tarafından geliştirilip devam ettirilir (Open-Source bunu en güzel örneği). Böyle kodların kim tarafından okunursa okunsun anlaşılır olması zorunludur. Aksi takdirde projeler, çalışmalar ve ürünler bugünkü hallerine ulaşamazdı.

#### Fonksiyon Tanımlama

Fonksyionların olayına değindik, şimdi fonksiyonları tanımlayıp da kullanma zamanı. Bir fonksiyon aşağıdaki yapıya sahiptir:
```python
def my_little_function(some_parameter, some_other_parameter):
    some code here
```
Burada **def** *define* kelimesinin kısaltılmışıdır, dolayısıyla tanımlamak için kullanmalıyız. Devamında fonksiyonun ismi gelir, isimlendirme kuralları değişkenler için olan kurallar ile aynıdır. Devamında ise parantezlerin içinde *parameter*'lar yer alır. Normalde bir fonksiyon hiç parameter almayabilir, ancak sonsuza kadar parameter alması mümkündür. Ancak programlama etiği (yazılı olmayan kurallar işte, uymazsanız Linus çok üzülür.) gereği bir fonksyionun **en fazla** 6 argument alması en doğrusudur. Fonskiyonun altında **içeride** onun kodu yer alır. Tıpkı döngüler ve koşullar gibi fonksiyonlar da iç içe tanımlanabilir.

#### Fonksiyon Çağırma

Fonksyion öyle durduğu yerde bekleyip ihtiyacımız olduğunu hissettiği vakitte devreye girmez. Onlara Batman gibi sinyal ışığı açmak lazım. Kullandığımız her yerde fonksiyonu çağırmak lazım. Python'da bu çok kolay:
```python
my_little_function(some_argument, some_other_argument)
```
Burada basit bir şekilde fonksiyonun ismini yazıyoruz ve içine *parameter*'lara karşılık gelecek şekilde *argument*'larımızı veriyoruz. Kelimeleri karıştırmayalım: Fonksiyonu tanımlarken yazıyorsak adı *parameter*, fonksiyonu çağırırken yazıyorsak adı *argument*!

````python
# Fonskiyon, fonksiyon ve daha çok fonksiyon!

def my_first_function(): # Gördüğünüz gibi argument almıyor!
    print("İlk fonksiyonum ve çok heyecanlıyım!")

for i in range(5):
    my_first_function() # Çağırmak da bu kadar basit!
````

````python
# İkinci bir fonksiyon

# Ben iyi isemlendiremedim siz daha iyi isimlendirin lütfen
def my_new_func(x,y):
    print(f"Sayıların çarpımı: {x} x {y} = {x*y}")

# Bunu yetişirse derste göreceğiz, yoksa bir sonraki derste
my_list = [x for x in range(0,20,2)]

# Saçmalıyorum burada maksat yeşillik olsun
for i in range(len(my_list)//2):
    my_new_func(my_list[i], my_list[-i-1]) # indexleri düzeltmek için -1 koyuyoruz çünkü ilk i 0 olacağı için -1 ile başlayabilmemiz lazım!
````

#### return Keyword

Bu fonksiyonlar biraz sıkıcı, çünkü sadece bir şeyler printliyorlar. Biz oysaki $f(x) = x + 5$ yazdığımızda bunun çıktısını kullanabiliyoruz rahatlıkla. Bunun için düşünmüşler taşınmışlar ve `return` diye bir *keyword* icat etmişler. `return` kendisinden sonra yazılan değeri/değişkeni (tek bir değişken dönmelidir!) fonksiyon bittiği sırada döner. Bu dönen değeri eğer bir değişkene eşitlersek artık hep kullanabiliriz!

````python
# Bu sefer güzel bir örnek buldum!

# Bir polinomun köklerini bulalım
def find_roots(a, b, c): # 2. dereceden olsun ax^2 + bx + c
    """
    Bu fonksiyon bir polinomun köklerini hesaplayıp döner.
    """

    delta = (b**2) - (4 * a * c) # b^2 - 4ac

    # İmajiner kökler
    if delta < 0:
        i = (-delta) ** 0.5 # 1/2 üssü kök demek ne de olsa!
        root1 = complex(-b/(2*a), i/(2*a)) # (-b ± kök(delta))/2a
        root2 = complex(-b/(2*a), -i/(2*a))

        # Birden fazla veri döeneceksek list, tuple, set ya da dictionary kullanabiliriz!
        # Tuple kullandım çünkü çıktım değişsin istemiyorum!
        return (root1, root2)

    # Tek kök
    elif delta == 0:

        # Birden farklı türde dönüş olabilir!
        # İki farklı return birbiri ile çakışmıyor o yüzden sorun yok!
        return -b/(2*a) # delta 0 olunca sadece a/2 olur tek kök!

    # Geriye tek olasılık kalıyor, o da iki reel kök
    else:
        root1 = (-b + (delta ** 0.5))/(2*a)
        root2 = (-b - (delta ** 0.5))/(2*a)

        return (root1, root2)

# 1, 2 ve 3. senaryo sırayla
my_list = [
    [5, -6, 5],
    [1, 4, 4],
    [1, -8, 15]
    ]

for li in my_list:
    result = find_roots(li[0], li[1], li[2])
    print(result)
````

### Part 2: Argument Türleri

Python 4 farklı argüman türüne sahiptir. Argümanlar bir fonksiyonun kalbidir. Bu yüzden hadi öğrenelim!

#### Varsayılan (*Default*) Argümanlar

Bazen bir fonksiyonda tüm argümanları sağlamak zorunda kalmayabiliriz. Bir fonksyionun zaten bizden beklediği bir değer olabilir, ve bu varsayımla hareket edebilir. Bunlar için *default* argümanlar bulunmakta! eğer bir argüman fonksyion tanımı sırasında `def foo(x, y=50):` şeklinde tanımlanırsa burada `y` *default* argument olarak anılır. Yani biz bir y değeri sağlamadığımız sürece fonksiyon y değerini 50 olarak alır. Verildiği takdirde o değeri kullanır.

#### Konumsal (*Positional*) Argümanlar

Bunu zaten aslında az önce hep kullandık. Fonksiyona parametreleri hep argümanları sırada (konumda) verdik. Python da bu parametreleri direkt o pozisyonda bulunan argüman ile otomatik olarak eşledi. Bunlar *positional* argument olarak adlandırılmakta.

#### *Keyword* Argümanları 

Bu ise biraz farklı. Bunu fonksiyon tanımlarken değil de çağırırken yaptığımız eylem etkiliyor. String formatlarken *.format()* sırasında istersek direkt *value* yazmak yerine *keyword* = *value* olarak kullanabiliyorduk. Böylece istediğimiz sırada *value*'ları sıralayabiliyorduk. İşte fonksyion çağırırken bunu birebir yapabiliyoruz! Örneğin `def bar(x, y):` diye tanımlanan bir fonksiyona parametreleri `bar(y=40, x=30)` şeklinde verebiliriz! Bunlar ise doğal olarak *keyword* argument olmuş olur.

````python
# Fonksyionlar kendi içinde fonksiyon tanımlayabilir ve çağırabilir!

def foo(x, y, z = 40): # z burada default argument!
    return x + y * z

def bar(x, y): # Burada argümanların tipini bilmiyorum, çağırırken söyleyebilirim sadece!

    # Burada keyword argument ile çağırdım!
    res1 = foo(z = 30, x = y, y = x)

    # Burada ise aynı fonksyionu positional argument ile çağırdım!
    res2 = foo(x, y) # z burada default!

    print(f"result 1: {res1}, result 2: {res2}")

bar(7, y = 3) # Bozmadığımız sürece ikisini beraber kullanabiliriz
````

#### *Arbitrary* Argümanlar

Bazen kaç adet argüman almamız gerektiği değişebilir. Normal şartlar altında fonksiyonu tanımlarken belirlediğimiz argüman sayısı dışında bir sayıda parametre verdiğimiz zaman hata alırız. Ancak *arbitrary* argümanları kullanırken hiçbir sorun yaşamayız, istediğimiz sayıda verebiliriz! İki tür bulunmaktadır: `*args` ve `**kwargs`. `*args` *arguments*'ın kısaltması olup bir liste şeklindedir, `**kwargs` ise *keyword arguments*'ın kısaltması olup dictionary şeklindedir!

````python
# Hemen örnekleyelim!
# İkisini beraber de kullanabiliriz!
def func(*args, **kwargs):
    print("Positional Arguments (*args):")
    for arg in args:
        print(arg)

    print("Keyword Arguments (**kwargs):")
    for key, value in kwargs.items():
        print(f"{key} - {value}")

func("Red", 'Hot', first='Chili', second='Peppers')
````

### Part 3: *Recursion* (Özyineleme)

Bir fonksiyon başka bir fonksiyonu çağırabilir. Peki kendini çağırabilir mi? Evet! Biz buna *recursion* diyoruz. Bir fonksiyon kendini çağırıp kendi üzerinde hesaplamalar yapabilir. Ancak burada sonsuz döngülere dikkat etmek lazım. Dolayısıyla bir *recursive* fonksyionun 3 temel parçası olmalıdır:

1. **Base Case:** Recursion işleminin artık durması gerektiğinde bunu tetikleyen koşuldur. Bu koşula ulaşıldığında fonksiyon geriye dönmeye başlar.
2. **Progress:** Fonksiyonun kendi iç işlemleridir.
3. **Recursive Call:** Fonksiyonun kendini çağırmasıdır.

*NOT: Base case her zaman bu üçlü içerisinde yapılmalıdır. 2 ve 3. kısımlar farklı şekillerde olabilir!*

````python
# En önemli örnek! Faktoriyel!

def factorial(num):

    # Base case, 0 ya da 1 ise hemen 1 dönmeliyiz!
    if num == 1 or num == 0:
        return 1
    
    # Recursive call ve progress burada birleşik! 
    # Önce recursion tamamlanır, sonra sayılar çarpılarak yukarı çıkar!
    return num * factorial(num-1)

print(factorial(5))
````

### Part 4: Variable Scoping

Fonksiyonlarla uğraşırken bazen değişken isimleri karışabiliyor birbirine. Aynı isimle bir değişken çeşitli yerlerde tanımlanabiliyor. Bir yerde erişilen bir değişkenin nereye ait olduğunu bulmanın bir yolu var! Python'da bir değişkenin nereye ait olduğu şu sıraya bakılarak bulunur:

1. **Local:** Fonksiyonun kendi değişkenleri.
2. **Enclosing:** Fonksiyonun (varsa) içinde bulunduğu fonksiyondan gelen değişkenler.
3. **Global:** Fonksyionların dışında yer alan değişkenler
4. **Built-in:** Python tarafından belirlenen isimler

````python
# Örnekle başlayalım!

# sum() fonksyionu Python'da built-in olarak bulunan bir fonksiyon,
# ancak Python bizim sum'ı değişken olarak kullanmamızı engellemiyor!

sum = 0 # Global değişken, artık sum() erişilemez!
print("global:", sum)

def foo():
    sum = 5 # foo() için local bar() için enclosing

    sum += 3 # hemen üstündeki (foo() için local) kullanılır

    print("enclosing:", sum)

    def bar():

        sum = 7 # local, foo() buna erişemez

        sum += 10 # local olan kullanılır

        print("local:", sum)

    bar() # local olanı çağırmak zorundayım kullanmak için!

    print("enclosing:", sum)

foo()
print("global:", sum)
````

### Part 5: Lambda Calculus

Lambda calculus bizim alışık olmadığımız bir matematik biçimi. Bugüne kadar okulda gördüğümüz matematikten çok daha ayrı, çok daha farklı bir şey. [Alonzo Church](https://en.wikipedia.org/wiki/Alonzo_Church) tarafından 1930'larda icat edilmiştir. Kendisi aynı zamanda Bilgisayar Biliminin kurucusu [Alan Turing](https://en.wikipedia.org/wiki/Alan_Turing)'in hocasıdır (çok büyük adamlar ikisi de). Lambda calculus'ta 3 temel şey bulunmaktadır:

1. **Değişkenler:** x, y, z gibi çeşitli değişkenler
2. **Fonksiyonlar:** $\lambda$ ile yazılan çeşitli fonksiyonlar, örneğin: $\lambda x. x + 1$
3. **Uygulamalar:** Değerlerin fonksiyonlar ve değişkenler üzerinde uygulanması, örneğin $(\lambda x. x + 1) 5$

Lambda calculus'ta her şey fonksiyon temellidir, mantıksal olarak kullandığımız *True*, *False*, *or*, *and* gibi kelimeler bile bir fonksiyon olarak ifade edilir. Daha fazlası için [buraya tıklayabilirsiniz!](https://en.wikipedia.org/wiki/Lambda_calculus) Günümüzde lambda calculus'tan ilham alınmış olan functional paradigm kullanan diller bulunmaktadır. Bunlardan en önemli olan dil [Haskell](https://en.wikipedia.org/wiki/Haskell)'dir. Günümüzde bankacılık uygulamalarının büyük bir kısmında, Meta'nın kötü amaçlı kullanım tespitlerinde, NASA'nın kritik uçuş sistemlerinde ve dünyanın en büyük dosya dönüştürücüsü [Pandoc](https://pandoc.org/)'ta Haskell kullanılmaktadır.

Python'da ve diğer functional paradigm içermeyen dillerde yine de lambda calculus'tan faydalanmaktayız! Python'da `lambda` kelimesini kullanarak basit *pseudofunction* tanımlayabiliriz! Örneğin $(\lambda x. x \times x)$ fonksiyonu bir $x$ alır ve onun karesini döner. Python'da ise bunu `square = lambda x: x * x` şeklinde yazıp gene fonksiyon gibi `square(5)` şeklinde kullanabiliriz! ':' işaretinin sol tarafında kalanlar (birden fazla değişken alabilir) *input* sağ tarafındaki ifadeyi ise *output* olarak kullanılır.

````python
# Örneklerle pekiştirelim!

# Pseudofunction'larımızı kullanalım!
square = lambda x: x ** 2 
square_root = lambda x: x ** (1/2) # 1/2'nci üs karekök ne de olsa!

print(square(25))
print(square_root(25))

````

````python
# Birden fazla değişken alan lambda'lar!

power = lambda x,y: x ** y
root = lambda x,y: x ** (1/y) 

print(power(3,4))
print(root(64,3)) # float'lar gene yaptı yapacağını,
# bunun düzgün versiyonunu meraklısına'da öğreneceğiz merak etmeyin

# Ternary alan lambda

is_even = lambda x: "Even" if x % 2 == 0 else "Odd"

print(is_even(15))
print(is_even(4))

# Bunu evde yapmayın! Sağlığınız için bakın ve geri unutun

fact = lambda x: x * fact(x-1) if x > 1 else 1

print(fact(5))
````

Python'da peki o zaman lambda'yı nasıl ve ne zaman kullanacağız? Python, ilginç özelliklere sahip bir dil. Kendi built-in fonksiyonları (meraklısına'da öğreneceğiz) ve bizim yazdığımız fonksiyonlarda argüman olarak bir fonksiyon verebiliriz! Çoğu zaman argüman olarak vereceğimiz fonksiyonlar basit olabiliyor, dolayısıyla her seferinde yeni bir fonksiyon tanımlamak yerine lambda ifadelerini argüman olarak vermek en doğru karar olur!

````python
# Örnek fonksiyon map()! 
# map(function, iterable) yapısına sahiptir, ilk argüman olarak bir fonksiyon alır, 
# bunu aldığı iterable'ın her item'ına tek tek uygular ve cevabı döner!

my_list = [10, 25, 50, 11, 24, 34, 42, 42, 17, 9, 81]

# Tek sayı ise üssünü aldı, çift sayı ise 3 ile çarptı!
my_list = list(map(lambda x: x ** 2 if x % 2 == 1 else x * 3, my_list)) # map() yaptıktan sonra list() ile listeye çevirmeyi unutmayın!

print(my_list)
````

````python
# Aynı şeyi map(), ternary ve lambda olmadan nasıl yazardık peki?

my_list = [10, 25, 50, 11, 24, 34, 42, 42, 17, 9, 81]

for item in my_list:
    # Tek ise
    if item % 2 == 1:
        # item'ı değiştirmek listemi etkilemiyor hatırlayalım!
        my_list[my_list.index(item)] = item ** 2
    else: # Çift ise
        my_list[my_list.index(item)] = item * 3

print(my_list)
# Bu sefer çok bir uzunluk farkı olmadı ancak farklı durumlarda lambda çok hayat kurtarıcı oluyor!
````

### Üçüncü Ders Bitti!
#### Evet, 3. dersin sonuna geldik! Bugün fonksiyonları ve lambda'yı öğrendik. Şimdi her şeyi anlamlı bir programda toplama vakti!

Bu sefer cos için [Taylor Estimation](https://en.wikipedia.org/wiki/Taylor_series) yapacağız! Bu sayede cos fonksiyonuna gerçek anlamda sahip olmadan belirli bir yakınlık değerinde bir sayının cos'unu bulabileceğiz. Cos için taylor estimation: $$\cos(x) \approx 1 - \frac{x^2}{2!} + \frac{x^4}{4!} - \frac{x^6}{6!} + \dots$$ diye gitmektedir. Tam formülü ise: $$\cos(x) = \sum_{n=0}^{\infty} \frac{(-1)^n x^{2n}}{(2n)!}$$ şeklindedir. Hadi bunu kullanarak fonksiyonumuzu yazalım!

````python
# Düzgün olan faktoriyeli kullanacağız!
def factorial(num):
    if num == 1 or num == 0:
        return 1
    return num * factorial(num-1)

PI = 3.141592653589793 # Pi tanımladık, normalde işin raconu 3 diyip geçmekttir de bu seferlik yapmadık!


def cos_taylor(num, degree): # Sonsuze kadar gidemeyeceğimiz içn bize bir degree lazım, kaçıncı dereceden estimation yaptığımızı göstermek için
    sum = 0
    num = num % 360 # 0-360 derece arasında tutalım              
    num = num * (PI / 180) # Sayıyı derece'den radyana çevirelim!
    for i in range(degree+1):
        neg_pow = (-1) ** i # (-1)^n
        num_pow = num ** (2*i) # x^2n
        fact = factorial(2*i) # (2n)!

        ith_step = (neg_pow * num_pow) / fact # i'nci basamaktaki sayı
        sum += ith_step

    return sum

# Hadi kullanalım!

# cos(10) = 0.86602540378 google hesap makinesine göre
print(cos_taylor(30,20)) # 20. dereceden estimate edelim!

# cos(60) =  0.5
print(cos_taylor(60,30)) # bu ise 30. dereceden!
````
