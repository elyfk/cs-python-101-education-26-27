### Part 1: Önemli Built-in Fonksiyonlar
Derste fonksiyonları tanıdık. Python bize özellikle listelerde kullanabilmemiz için birçok built-in fonksiyon da vermiş. Hadi önemli olanları beraber tanıyalım!

#### Listeler İçin Önemli Built-in Fonksiyonlar

* `len(list)`: Bunu önceki derslerimizde işlemiştik, ancak tekrardan burada yer vermek lazım. Listenin eleman sayısını döner.
* `sum(list, start=0)`: Listedeki elemanların toplamını döner, opsiyonel olarak `start=` ile bir index de verilebilir!
* `min(list)` ve `max(list)`: İlki verilen listedeki en küçük elemanı, diğeri ise en büyük elemanı döner.
* `reversed(list)`: Listeyi tersten sıralayıp döner
* `sorted(list, key=function, reverse=False)`: Listeyi sıralar, `reverse=` değerine göre tersten sıralayabilir, opsiyonel olarak `key=` kısmına aldığı fonksiyon eşliğinde fonksyiondan gelen değerler ile listeyi sıralayabilir.
* `map(function,list)`: Verilen fonksiyonu tüm elemanlara tek tek uygular.
* `filter(function,list)`: Verilen fonksiyon sonucunda *True* değeri dönen değerleri listede tutar, gerisini atar.
* `any(list)`: Listenin en az bir elemanı *True*'ya denk ise *True* değilse *False* döner.
* `all(list)`: Listenin tüm elemanları *True*'ya denk ise *True*, biri bile *False*'a denk ise *False* döner.

#### Matematiksel Built-in Fonksiyonlar

* `abs(number)`: İçine aldığı sayının mutlak değerini döner.
* `round(float, precision)`: İçine aldığı float'ı yuvarlar, ikinci argüman verilmezse integer, verilirse vigülden sonra o kadar basamak gidip yuvarlar.
* `pow(base, exp)`: İçine aldığı 1. sayının 2. sayı kadar üssünü alır.
* `divmod(num1, num2)`: İçine aldığı sayılar için bir tuple döner, tuple formatı `(a // b, a % b)` işlemlerinin cevabı şeklindedir.

````python
# Deney vakti!

my_list = ["Zeynep", "Ahmet", "Arda", "İzzet", "Erenalp", "Bilgehan", "Nisa", "Mine", "Damla", "Selen"]

print("List length:", len(my_list)) # Kaç eleman var görelim!

# Her elemanın uzunluğunu görelim
print("Length of each name:", list(map(lambda x: len(x), my_list))) # map() ve filter() çıktılarını list() ile dönüştürmeyi unutmayın!

print("Only names longer than 4 caharacters:",list(filter(lambda x: len(x) > 4, my_list))) # 4 Harften uzun isimleri alalım

sort_list = sorted(my_list, key=len) # uzunluklarına göre sıraladım!

print("First sort:", sort_list)

# Önce isimleri uzunluğuna göre sıraladık, sonra aynı uzunluktaysa alfabetik sıraladık!
sort_list_complex = sorted(my_list, key= lambda x: (len(x), x.encode('utf-8'))) # lambda da parantez ile birden fazla kriter verebiliriz!
# .encode() ile harfleri önce sayıya dönüştürdük, böylece kıyaslayabildik!

print("Second sort:", sort_list_complex)
````

### Part 2: List Comprehension

Python'da bazen listelere ihtiyacımız olduğunda şu ana kadar hep elle tek tek yazdık. Bunu aslında belirli bir seviyeye kadar otomatikleştirebiliriz! Bunun için ise *list comprehension* var! Bu sayede elimizde bir liste varso ondan daha hızlı bir liste, yoksa da yeni listeleri çok daha hızlı üretebileceğiz! Bunun için Python'da `my_list = [x for loop]` şemasını kullanıyoruz.

````python
# Örnekler ile daha iyi anlayacağız!

my_list = [x for x in range(20)] # 0'dan 19'a bir liste oluşturduk anında

print(my_list)
````

````python
# Başka bir listeden oluşturalım!

my_list = ["Zeynep", "Ahmet", "Arda", "İzzet", "Erenalp", "Bilgehan", "Nisa", "Mine", "Damla", "Selen"]

my_new_list = [x.upper() for x in my_list] # Herkesi büyük harf yaptık!

print(my_new_list)
````

````python
# if-else de kullanabiliriz!
# for'a göre sol tarafta kullanırsak x üzerinde if-else kullanırız.
# for'a göre sağ tarafta kullanırsak if'i sağlayan x'leri seçmiş oluruz.

my_list = [x for x in range(100) if x % 7 == 0] # Burada else koyamayız, sadece 7'ye tam bölünen x'leri seçer

print(my_list)

my_list = [x/2 if x> 10 else x * 3 for x in range(20)] # Burada seçtiğimiz x'ler üzerinde işlem yaparız!

print(my_list)
````

#### Math Library

Math *library* Python tarafından hazırlanmış en önemli *library*'lerden biridir. Bu library bir çok sayı ve fonksiyona sahiptir, hadi en önemlilerini keşfedelim!

#### Sabit Sayılar
* `e`: Euler sayısı
* `pi`: Pi sayısı
* `tau`: Tau sayısı (Pi'nin iki katıdır)

#### Numerik Fonksiyonlar
* `ceil(float)`: Sayıyı üste yuvarlar
* `floor(float)`: Sayıyı alta yuvarlar
* `factorial(num)`: Faktoriyel hesaplar
* `gcd(num1, num2)`: İki sayının EBOB'unu (En Büyük Ortak Bölen) hesaplar
* `fabs(num)`: Sayının mutlak değerini alır ve **float** olarak döner

#### Logaritmik ve Üstel Fonksiyonlar
* `exp(num)`: $e^x$ değerini hesaplar
* `pow(num1, num2)`: $x^y$ değerini hesaplar
* `log(a, b)`: $\log_{b}a$ değerini hesaplar
* `log2(num)`: $\log_{2}x$ değerini hesaplar
* `log10(num)`: $\log_{10}x$ değerini hesaplar
* `sqrt(num)`: $\sqrt{x}$ değerini hesaplar

#### Trigonometrik Fonksiyonlar
* `sin(num)`, `cos(num)` ve `tan(num)`: Verilen sayıya göre hesaplarlar
* `degrees(num)`: Radyandan dereceye dönüştürür
* `radians(num)`: Dereceden radyana dönüştürür

Tabii ki daha birçok fonksiyon bu *library*'de yer alıyor. Devamını keşfetme ise size kalmış!

````python
# Dersteki taylor estimation ile gerçeğini kıyaslayalım!

from math import cos, pi, radians

# ---- Dersteki Kod ----
def factorial(num):
    if num == 1 or num == 0:
        return 1
    return num * factorial(num-1)


def cos_taylor(num, degree): # Sonsuze kadar gidemeyeceğimiz içn bize bir degree lazım, kaçıncı dereceden estimation yaptığımızı göstermek için
    sum = 0
    num = num % 360 # 0-360 derece arasında tutalım              
    num = num * (pi / 180) # Sayıyı derece'den radyana çevirelim!
    for i in range(degree+1):
        neg_pow = (-1) ** i # (-1)^n
        num_pow = num ** (2*i) # x^2n
        fact = factorial(2*i) # (2n)!

        ith_step = (neg_pow * num_pow) / fact # i'nci basamaktaki sayı
        sum += ith_step

    return sum

# ---- Dersteki Kod ----

# Çıktıları kıyaslayalım!
def compare(cosine, taylor):
    percentage = (taylor / cosine) * 100
    print(f"The estimated value is the {percentage} of the real value, the error is {(1 - (percentage/100)):.16f}")


compare(cos(radians(60)), cos_taylor(60, 5)) # 5. dereceden estimation 60 derece için

compare(cos(radians(30)), cos_taylor(30, 1)) # 2. dereceden estimation 30 derece için

compare(cos(radians(45)), cos_taylor(45, 15)) # 15. dereceden estimation 45 derece için
````
