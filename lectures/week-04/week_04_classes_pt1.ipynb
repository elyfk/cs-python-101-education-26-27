### Part 1.1: OOP'ye Giriş

Şuana kadar elimizde tuttuğumuz değişkenler ve fonksiyonlar kodun orasında burasında duruyordu. Bir ürün, bir nesne, bir cihaz hakkındaki verileri ve işlemleri tutmak istediğimizde her şey çok karışık bir şekilde halledilmiş oluyor. Bu sorunu gidermek için ise *Object Oriented Programming (OOP)* var! OOP sayesinde bir nesneye yönelik tüm verileri ve işlemleri tek bir yerde tutabiliyoruz!

OOP'de nesneler için kullandığımız genel şemaları (Template) *Class*, o *class*'ın her örneğini, her kopyasını ise *object* olarak adlandırıyoruz! Her *class* içerisinde belirli değişkenlere ve fonksiyonlara yer verebiliriz! Bu *class*'ı kullanarak ürettiğimiz her *object*'te ise bu değişkenleri nesne'nin kendisine göre ayarlayabiliriz ve yer alan fonksiyonları kullanabiliriz!

Hadi bir örnek ile *class*'ları anlayalım!
```python
class Animal:
    def __init__(self, name, sound, weight, species):
        self.name = name
        self.sound = sound
        self.weight = weight
        self.species = species
```

Yukarıdaki örnekte olduğu bir *class* tanımlayabilmek için `class` kelimesini kullanmamız gerekmektedir. Devamında o *class*'ın ismi yer alır ve *class* tanımlanmış olur. *Class* içerisinde yer alan değişkenler *attribute* olarak adlandırılır. Her oluşturulan *object* aynı *attribute*'lara sahiptir. Örnekte de olduğu gibi *class*'lar fonksiyonlara sahip olabilir, bu fonksiyonlar ise *method* olarak adlandırılır (şimdi taşlar yerine oturdu). Python her *class*'ta kullanılması için bazı *method*'ları kendisi tanımlar. Bunlardan en önemlisi ise `__init__` *method*'udur ve *initialize* kelimesinin kısaltılmışıdır. Bu *method* sayesinde bizler her *class*'tan *obkect* ürettiğimizde *attribute*'larını belirlemiş oluruz. Örnekte görüldüğü gibi ilk argüman `self`, peki bu ne anlama geliyor? `self` kelimesi *object*'in kendisini işaret eder. Onun sayesinde her nesnenin değişkenlerine erişebilir ve kullanmamıza olanak tanır. *Object*'lerin *attribute*'ları `self` ile tanımlanmak **zorunda** olup normalde `__init__` içerisinde tanımlanması **zorunlu değildir**. Kod anlaşılırlığı açısından boş olarak tanımlansa bile `__init__` içerisinde tüm *attribute*'ların tanımlanması doğru olandır.

Peki tanımladığımız bu örnek *class*'tan nasıl yeni *object*'ler üretebilirim?
```python
cat = Animal("Java", "Meow", 5, "Cat") # CENG'in kedisi Java vardı maalesef vefat etti :(
```

Yukarıdaki örnekte daha önce tanımladığımız `Animal` adlı *class*'tan bir *object* üretmiş ve bunu cat değişkenine atamış olduk! Bu oluşturma sırasında `__init__` otomatik olarak çağrılmış ve *attribute*'lar yerleştirilmiş oldu. Ancak `self` kayboldu! Evet, self sadece tanımlama sırasında kullanıldı, *class* tanımlanması dışında `self` *asla* argüman olarak kullanılmaz, Python onu otomatik halleder. Ek olarak tanımlanan **her *method* için** `self`'in ilk argüman olması **zorunludur**! `self` *class* yapılarında olmazsa olmaz ve oldukça gerekli bir kelimedir!

````python
# Güzel bir class oluşturalım!

class Soldier:
    """
    This is my soldier implementation! 
    It's all up to your imagination!
    """
    def __init__(self, name: str, age: int, faction: str, type: str, hp: float, stamina: float): # Type hinting sadece yazılımcı içindir unutmayalım!
        self.name = name
        self.age = age
        self.faction = faction
        self.type = type
        self.hp = hp
        self.stamina = stamina

# İlk askerimi oluşturdum!
my_soldier0 = Soldier("Ahmet", 20, "Khuzait", "Cavalry", 200.0, 150.0)

# Print ne diyecek acaba?
print(my_soldier0) # korkmayın, bunu yakında düzelteceğiz! Şuanda aldığınız çıktı sadece onun hafızada tutulduğu yeri gösteriyor!

# Tipine de bi bakalım!
print(type(my_soldier0)) # __main__ kısmı bizim şuanda bulunduğumuz dosyada yer aldığını ifade etmekte!

# Attribute printleyelim!
# Attribute'lara . ile ulaşabiliyoruz!
print(my_soldier0.name)
````

### Part 1.2: Attribute'ları Anlamak

Attribute'lar Python'da bizim object'lerin sahip olduğu özellikleri anlamamıza yarar. Bir class yazmaya karar verdikten sonra yapmak gereken 3 adım vardır: İyi attribute'lar belirlemek, işlevsel method'lar tanımlamak ve hayatı sorgulamak. Şu anda neyse ki ilk adımdayız! İyi bir attribute şu özelliklere sahip olmalı:

* Attribute'lar bir işlem ya da bir davranış belirtmemeli. İyi bir attribute, ait olduğu nesnenin bulunduğu bir durumu ya da bir özelliğini ifade etmeli. Örneğin, bir araba için `color` iyi bir attrbute olur, ancak `is_moving` kötü bir attribute olur, çünkü hareket etmek arabanın bir davranışıdır, hafızada tutulması gereken bir özellik değil.
* Attribute'lar genel olmalıdır. Bir class'a ait attribute, o class'ın object'lerini birbirinden ayırt etme konusunda işlevsel olmalıdır. Genellenebilir attribute'lar her zaman spesifik olanlardan iyidir. Örnekle pekiştirmek gerekirse bir araba için `color` gerçekten de iyi bir attribute iken `is_yellow` genel bir ayraç işlevi görmediği için kötü bir attribute olur.
* Attribute'lar class ile alakalı olmalıdır. Class'a doğrudan ait olmayan attribute'lar veri yönetiminde sorunlara yol açar. Örneğin bir araba için `color` bayağı iyi bir attribute iken `drivers_licence` alakasız kaldığı ve arabadansa arabanın sürücüsüne ait olması gereken bir attribute'dur.
* Attribute'lar gerekli olmalıdır, bir class altında birçok attribute yazılması mümkünken program dahilinde gerekli olan ve ileride gerekli olacağı öngörülen veriler dışında gereksizdir. Örneğin bir arabada `color` inanılmaz iyi bir attribute iken `door_handle_count` diye bir attribute'a sahip olmak cidden kimsenin işine yaramaz!
* Attribute'ların kime ait olduğu iyi belirlenmelidir. Attribute'un class'ın kendisine mi ait olması gerektiği yoksa sadece object'lerin hepsinde bulunması gerektiği belirtilmelidir. Bir dakika, class'a ait attribute mu var, hepsi object'lerin değil miydi? Tabii ki de hayır! Hemen oraya gelelim!

2. **Class Attribute'ları:** Bu tür attribute'lar ise `self` kelimesi olmadan tanımlanır. Class içerisindeki tüm object'ler için erişilebilir, ve tüm object'ler için sabittir.

### Part 1.3: Method'ları Anlamak

Method'lar Python'da sahip olduğumuz object'lerin davranışlarını anlamamızı sağlar. Class yazmanın 2. aşamasında yer alır. İyi bir method şu özelliklere sahip olmalı:

* Method'lar anlaşılır ve yaptığı işlemi anlatıcak şekilde isimlendirilmelidir. İyi bir method, yaptığı işlemi 2-3 kelimede anlatabilmelidir. Örneğin bir arabada `calculate_price` güzel bir method iken `calculate_the_price_of_the_car` okuması uzun ve her seferinde tekrarlanması gereken bir isim olup mkötü method örneğidir.
* Method'lar tek bir işlemi yapması gerekmektedir. Standard, belli bir işlemi yapan ve bunu iyi yapan bir method iyi bir method'dur. Örneğin bir arabanın attribute'larından yola çıkıp fiyatına hesaplayan bir method oldukça güzel iken bir arabanın benzin fiyatı üzerinden kilometre başına harcadığı parayı hesaplayıp bunun yanı sıra araba kilometre yaptıkça fiyatını otomatik düzenleyen tek bir method fazla kalabalık olup kötü bir method'dur.
* Method'lar ait olduğu class'a odaklı olmalıdır. Bir method ya class'ın attribute'larını okuyup onlardan bir sonuç üretmeli, ya class'ın attributelarını modifike etmeli ya da ikisini birden yapabilmelidir. Örneğin bir arabanın attribute'larından yola çıkıp fiyatına hesaplayan bir method aşırı güzel iken arabanın sürücüsünün ödemesi gereken cezaları hesaplayan bir method arabaya değil sürücüye aittir.
* Method'lar tıpkı attribute'lar gibi gerekli ve genel olmalıdır. Örneğin bir arabada `calculate_price` inanılmaz bir method iken `is_yellow` adlı bir method oldukça gereksizdir ve anlamlı bir işlev ortaya koymaz.
* Method'ların kime ait olduğu iyi belirlenmelidir. Method'un türü her zaman açıkça belirlenmelidir. Bir dakika, burada da mı çeşitler var? Evet hem de 3 çeşit mevcut! Hemen oraya gelelim!

#### Method Türleri

1. **Instance Method'ları:** Class içerisinde normal bir şekilde tanımlanan method'lardır. Object'i temsil eden `self` kelimesini ilk argümanı olarak alır. Her object için ayrı olarak çalışır.
2. **Class Method'ları:** Class attribute gibi tüm class'ı kapsar. Class'ı temsil eden `cls` kelimesini ilk argümanı olarak alır. `@classmethod` decorator (ne olduğunu öğreneceğiz) ile tanımlanır.
3. **Static Method'ları:** Class ya da object ile alakalı değildir, daha çok fonksiyonları gruplamak için kullanılır, `cls` ya da `self` argümanı almaz. `@staticmethod` ile tanımlanır.

````python
# Yeni bir class inşa edelim!

class Faction:
    """
    This is my faction implementation! 
    It's all up to your imagination!
    """
    # Tüm faction'lara erişmek istiyorum
    all_factions = []
    
    @classmethod
    def list_factions(cls):
        return cls.all_factions

    def __init__(self, name, lands, troops, ruler, enemies = [], friends = []):
        self.name = name
        self.lands = lands
        self.troops = troops
        self.ruler = ruler
        self.enemies = enemies
        self.friends = friends
        Faction.all_factions.append(self) # Böylece kendimi ekledim


    def can_hire(self, soldier: Soldier):
        if soldier.faction != self.name:
            return False
        
        if soldier.hp >= 100.0 and soldier.stamina >= 200.0:
            return True
        elif soldier.hp >= 250.0 and soldier.stamina >= 100.0:
            return True
        else:
            return False

my_new_faction = Faction("Aserai", [], [], "Erenalp")

print(my_new_faction) # korkmayın, bunu yakında düzelteceğiz! Şuanda aldığınız çıktı sadece onun hafızada tutulduğu yeri gösteriyor!

# Class method'u test edelim!
print(Faction.list_factions()) # korkmayın, bunu yakında düzelteceğiz! Şuanda aldığınız çıktı sadece onun hafızada tutulduğu yeri gösteriyor!

# Bu askerleri alabiliyor muyum görelim!
print(my_new_faction.can_hire(my_soldier0))

print(my_new_faction.can_hire(my_soldier1))
        
````

````python
# Static method'lar ile bir class yapalım!

class Tax:
    """
    This is my tax implementation! 
    It's all up to your imagination!
    """

    @staticmethod
    def soldier_tax(soldier: Soldier):
        """
        Calculates a single soldier's tax
        """

        match soldier.type:
            case "Cavalry" : return 250
            case "Archer" : return 125
            case "Halberd": return 175
            case "Swordsman": return 100
            case _: return 0

    @staticmethod
    def faction_tax(faction: Faction):
        """
        Calculates the whole faction's soldiers' taxes
        """

        tax = 0
        for troop in faction.troops:
            tax += Tax.soldier_tax(troop)

        return tax

# Yeni askerler yaptım!
my_soldier2 = Soldier("Samet", "Aserai", "Halberd", 280.0, 100.0)
my_soldier3 = Soldier("Kenan", "Aserai", "Catapultsman", 300.0, 160.0) # Bunun türü yok! O yüzden 0 dönecek!

my_new_faction.troops.append(my_soldier1)
my_new_faction.troops.append(my_soldier2)
my_new_faction.troops.append(my_soldier3)

print(Tax.soldier_tax(my_soldier2))
print(Tax.faction_tax(my_new_faction))
````

### Part 1.4: Dunder'ları Anlamak

Dunder'lar (Double Underscore'un kısaltmasıymış saat 00.40'ta öğrenmek de ilginç oldu.) yukarıda kullandığımız `__init__` ve onun gibi hem başı hem de sonu '__' olan fonksiyonlardır. Bu fonksiyonlar class'larda çeşitli özel işlevleri yerine getirir, bunlar sayesinde oldukça ilginç eylemler ve özelleştirmeler yapabiliriz! Python'da en çok kullandığımız dunder'lar şunlardır:

#### Genel Dunder'lar
* `__init__(self)`: Zaten bahsettik, object oluşturulmasını ve başlatılmasını (initialise) sağlar.
* `__del__(self)`: Object yok edildiği zaman çağırılır. Normalde drekt yazılmasa bile otomatik çalışır, ancak özel işlemler için tanımlamak en doğrusudur
* `__abs__(self)`: abs() fonksiyonunun çağırması ile çalışır.
* `__round__(self)`: round() fonksiyonunun çağırması ile çalışır.
* `__neg__(self)`: Class'ın -object işlevini görür.
* `__pos__(self)`: Class'ın +object işlevini görür.
* `__len__(self)`: len(object) işelminin sonucunu tanımlar.

#### Aritmetik Dunder'lar
* `__add__(self,other)`: Object + other_object işlemini tanımlar.
* `__sub__(self,other)`: Object - other_object işlemini tanımlar.
* `__mul__(self,other)`: Object * other_object işlemini tanımlar.
* `__truediv__(self,other)`: Object / other_object işlemini tanımlar.
* `__mod__(self,other)`: Object % other_object işlemini tanımlar.

#### Printleme ve Benzeri İşlemlerde Otomatik Çağırılan Dunder'lar
* `__str__(self)`: str(object) işleminin sonucunu tanımlar. (Genelde dış kullanıcı içindir)
* `__repr__(self)`: repr(object) işleminin sonucunu tanımlar.
* `__bool__(self)`: bool(object) işleminin sonucunu tanımlar.

#### Karşılaştırma Dunder'ları
* `__eq__(self,other)`: Object == other_object işlemini tanımlar. (equal)
* `__ne__(self,other)`: Object != other_object işlemini tanımlar. (not equal)
* `__lt__(self,other)`: Object < other_object işlemini tanımlar. (less than)
* `__le__(self,other)`: Object <= other_object işlemini tanımlar. (less or equal)
* `__gt__(self,other)`: Object > other_object işlemini tanımlar. (greater than)
* `__ge__(self,other)`: Object >= other_object işlemini tanımlar. (greater or equal)


````python
# Eski class'ımız için Dunder tanımlayalım!

class Faction:
    """
    This is my faction implementation! 
    It's all up to your imagination!
    """
    # Tüm faction'lara erişmek istiyorum
    all_factions = []
    
    @classmethod
    def list_factions(cls):
        return cls.all_factions

    def __init__(self, name, lands, troops, ruler, enemies = [], friends = []):
        self.name = name
        self.lands = lands
        self.troops = troops
        self.ruler = ruler
        self.enemies = enemies
        self.friends = friends
        Faction.all_factions.append(self) # Böylece kendimi ekledim


    def can_hire(self, soldier: Soldier):
        if soldier.faction != self.name:
            return False
        
        if soldier.hp >= 100.0 and soldier.stamina >= 200.0:
            return True
        elif soldier.hp >= 250.0 and soldier.stamina >= 100.0:
            return True
        else:
            return False

    # Güzel bir kodlama pratiği olarak __init__ ilk method, diğer dunderlar ise en sonda tanımlanması tavsiye edilir
    def __str__(self):
        return f"{self.ruler}'ın Büyük {self.name} Krallığı"

my_new_faction = Faction("Aserai", [], [], "Erenalp")

print(my_new_faction) # Artık düzgün gösterecek!
````

#### `__str__` vs. `__repr__`

Bu iki dunder'ın aslında yaptığı şey aynı, object'i açıklayan bir string dönmek. Peki neden ikisi de var? Çünkü amaçları farklı. `__str__` genel olarak dış kullanıcıya hitap eder. Bir print sırasında `print(object)` yapıldığında Python gizlice `str(object)` çağırır (bunu aslında her zaman yapıyor) ve böylece object'i ekrana yansıtır. `__repr__` ise yazılımcı içindir. Debug ve benzeri kodu anlamayı gerektiren durumlarda kullanılır. İkisi farklı string'ler gösterecek şekilde yazılır. 
```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        # Unambiguous representation for developers
        return f"Book(title={self.title!r}, author={self.author!r})"

    def __str__(self):
        # Friendly display for end users
        return f"'{self.title}' by {self.author}"
```
Örneğinde `book = Book("Dune", "Frank Herbert")` object'i `__str__` için *'Dune' by Frank Herbert*, `__repr__` için ise *Book(title='Dune', author='Frank Herbert')* döndürür. Böylece hem kullanıcı hem de yazılımcı object'i kendi düzeylerinde anlamış olur. Peki ikisi ne zaman çağırılıyor, hangi eylem `__str__` ile hangisi `__repr__` ile etkileşime geçiyor? Hadi öğrenelim!

````python
# Örnekleyelim!

class Soldier:
    """
    This is my soldier implementation! 
    It's all up to your imagination!
    """
    soldier_count = 0
    
    def __init__(self, name: str, faction: str, type: str, hp: float, stamina: float):
        self.name = name
        self.faction = faction
        self.type = type
        self.hp = hp
        self.stamina = stamina
        Soldier.soldier_count += 1

    def __str__(self):
        return f"{self.faction}'den {self.name} adlı asker"

    def __repr__(self):
        return f"Soldier(name={self.name}, faction={self.faction}, type={self.type})"

my_soldier0 = Soldier("Ahmet", "Khuzait", "Cavalry", 200.0, 150.0)
my_soldier1 = Soldier("Arda", "Aserai", "Archer", 100.0, 200.0)
my_soldier2 = Soldier("Samet", "Aserai", "Halberd", 280.0, 100.0)
my_soldier3 = Soldier("Kenan", "Aserai", "Catapultsman", 300.0, 160.0)

print(my_soldier3) # __str__ çağırılacak

print([my_soldier0, my_soldier1, my_soldier2, my_soldier3]) # __repr__ çağırılacak
````
