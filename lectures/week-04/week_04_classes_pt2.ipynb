### Part 2.1: Inheritance

Inheritance, bir class'tan (parent) yeni class'lar üretebilme (child) becerisidir. Bazen bir ya da birden fazla class'ı kullanıp üstüne yeni bir class inşa etmemiz gerekebilir. Inherit ettiğimiz classların üstüne yeni attribute ve method ekleyebiliriz ya da parent'ı kullanabiliriz! Inherit etmek oldukça kolay, hadi yapalım!
```python
class Parent:
    pass

class Child(Parent):
    pass
```
Yukarıda olduğu gibi, bulunan bir class'ı parantez içinde yazarak child class içinde inherit edebiliriz. Normal kullanıldığı takdirde parent dahilinde yer alan her şey anında direkt child içinde çalışır hale gelir (buna `__init__` Python'a ait *dunder*'lar dahildir.), sadece **private** değerler aktarılamaz. Ancak değiştirmek istediğimiz bir şey vara aynı fonksiyonu yeniden tanımlayabiliriz. Aksi takdirde parent içerisinde dahil olan her method ve attribute otomatik olarak aktarılır.
```python
class Parent:
    def foo():
        pass

class Child(Parent):
    def bar():
        pass
```
Örneğinde `child.foo()` kullanıma açık olup birebir aynı işlevi görür. 
```python
class Parent:
    def foo():
        pass

class Child(Parent):
    def foo():
        pass
```
Örneğinde ise `child.foo()` ve `parent.foo()` birbirinden farklı çalışır, ikisi de kendine ait olanı kullanır. Buna *function override* denir.

*NOT: `__init__` override edildiğinde parent dahilinde yer alan attribute'lar kaybolur, onu yeniden tanımlamak gerekir!*

````python
# Yeni class üretelim
from enum import Enum

class SoldierType(Enum):
    MAGE = "Mage"
    CAVALRY = "Cavalry"
    ARCHER = "Archer"
    SWORDSMAN = "Swordsman"

class Soldier:
    """
    This is my soldier implementation! 
    It's all up to your imagination!
    """
    soldier_count = 0
    
    def __init__(self, name: str, faction: str, type: SoldierType, hp: float, stamina: float):
        self.name = name
        self.faction = faction
        if not isinstance(type, SoldierType):
            raise TypeError("It should be SoldierType")
        self.type = type
        self.hp = hp
        if hp > self.calculate_max_hp() or hp < 0:
            raise ValueError("Hp can't be lower than zero or greater than the max")
        self.stamina = stamina
        if stamina > self.calculate_max_stamina() or stamina < 0:
            raise ValueError("Stamina can't be lower than zero or greater than the max")
        Soldier.soldier_count += 1

    def calculate_max_hp(self):
        match self.type:
            case SoldierType.MAGE: return 175
            case SoldierType.CAVALRY: return 250
            case SoldierType.ARCHER: return 125
            case SoldierType.SWORDSMAN: return 200
            case _: return 0  

    def calculate_max_stamina(self):
        match self.type:
            case SoldierType.MAGE: return 300
            case SoldierType.CAVALRY: return 200
            case SoldierType.ARCHER: return 200
            case SoldierType.SWORDSMAN: return 150
            case _: return 0              
    
    def __str__(self):
        return f"{self.faction}'den {self.name} adlı asker"
    
    def __repr__(self):
        return f"Soldier(name={self.name}, faction={self.faction}, type={self.type})"


class Mage(Soldier): # Şuan ondaki her şeye sahibim!
    """
    This is my mage implementation! 
    It's all up to your imagination!
    """
    mage_count = 0

    def __init__(self, name: str, faction: str, hp: float, stamina: float, magic: float):
        self.name = name
        self.faction = faction
        # Mage sabitlenmiş oldu!
        self.type = SoldierType.MAGE 
        self.hp = hp
        if hp > self.calculate_max_hp() or hp < 0:
            raise ValueError("Hp can't be lower than zero or greater than the max")
        self.stamina = stamina
        if stamina > self.calculate_max_stamina() or stamina < 0:
            raise ValueError("Stamina can't be lower than zero or greater than the max")
        self.magic = magic
        if magic < 0:
            raise ValueError("Magic can't be lower than zero")
        Soldier.soldier_count += 1
        Mage.mage_count += 1    

    # Diğer fonksiyonları tekrar yazmama gerek yok!  


    # Bunu yazmam gerekli, çünkü farklı olması lazım!    
    def __str__(self):
        return f"{self.faction}'den {self.name} adlı büyücü"
        
    def __repr__(self):
        return f"Mage(name={self.name}, faction={self.faction}, magic={self.magic})"

# my_mage0 = Mage("Ahmet", "Khuzait", -80.0, 150.0, 150) # Hata verir! Çünkü hp negatif
# my_mage0 = Mage("Ahmet", "Khuzait", 480.0, 150.0, 150) # Hata verir! Çünkü hp max'tan daha fazla

my_mage1 = Mage("Arda", "Aserai", 100.0, 100.0, 150)
my_mage2 = Soldier("Arda", "Aserai", SoldierType.MAGE, 100.0, 100.0) # Bu mage class'ına ait değil, o yüzden sayılmıyor!

print(Mage.mage_count) # 1 vermeli
print(Soldier.soldier_count) # 2 vermeli
````

````python
# Erişimleri test edelim!

print(my_mage1.calculate_max_hp()) # Parent fonksiyonuna erişiyor

# print(my_mage2.magic) # Child'ında yer alan farklı attribute ve method'lara sahip değil!

# Aitliği test edelim!

print("--Child--")
# Child class'ın object'leri parent'a da aittir!
print(isinstance(my_mage1,Soldier))
print(isinstance(my_mage1,Mage))

print("--Parent--")
# Parent ise doğal olarak child'a ait değildir!
print(isinstance(my_mage2,Soldier))
print(isinstance(my_mage2,Mage))

print("--Subclass--")
print(issubclass(Mage, Soldier)) # Doğal olarak doğru
print(issubclass(Soldier, Mage)) # Doğal olarak da yanlış, ne beklediniz ki?
````

#### super()

Parent class'ı inherit ettikten sonra override ettiklerimizi baştan sona yazmak, hele ki `__init__`'i tekrardan yazmak çok zor değil mi? Bayağı üşendirici! O yüzden müthiş bir icat var elimizde: `super()`! `super()` sayesinde parent fonksiyonumuzu override ettiğimiz fonksiyon içerisinde çağırabiliyoruz!

*ÖNEMLİ NOT: `*args` ve `**kwargs` ile super() çağırırken yeni argümanları ya `*args`'tan hemen önce positional olarak ya da `*args` ile `**kwargs`'ın tam ortasında keyword olarak vermek gerekir.*

Aşağıdaki örnek için hem `def __init__(self, magic: float, *args, **kwargs):` hem de `def __init__(self, *args, magic: float = 200.0, **kwargs):` çalışır. 2. kısımda keyword argument formatında olması **zorunludur.** Dolayısıyla default value verilmiş olur. Eğer default value verilmek istenmiyorsa `*args`'tan önce verilmelidir, yani 1. kısım tercih edilmelidir!

````python
# Mage'i düzenleme vakti!

class Mage(Soldier): 
    """
    This is my mage implementation! 
    It's all up to your imagination!
    """
    mage_count = 0

    def __init__(self, magic: float, *args, **kwargs):
        # Böylelikle fazla şeyle uğraşmadık!
        # Sadece type olan kısım için araya MAGE koyduk (2. indexteydi Soldier için)
        # Normalde argüman sıralamayı doğru yaparsak bu tür şeyler yaşanmaz!
        super().__init__(*args[:2], SoldierType.MAGE, *args[2:], **kwargs)
        
        self.magic = magic
        if magic < 0:
            raise ValueError("Magic can't be lower than zero")
        Soldier.soldier_count += 1
        Mage.mage_count += 1    

    def __str__(self):
        return f"{self.faction}'den {self.name} adlı büyücü"
        
    def __repr__(self):
        return f"Mage(name={self.name}, faction={self.faction}, magic={self.magic})"

my_mage3 = Mage(150, "Görkmen", "Svadya", 100.0, 100.0) # Magic en başa geldiği için sıralama oynadı tabii ki!
````

### Part 2.2: Encapsulation

Encapsulation ise sahip olunan attribute'lara ve method'lara kimin erişip erişemeyeceğine karar vermek demektir. Bazen bir attribute'u kontrol etmek ya da bir fonksiyonu herkesin kullanımına imkan tanımak gerekebilir. Bunun için ise üç adet attribute ve method türü bulunmaktadır:

1. **public:** Dışarıdan kullanıma tamamen açıktır. Şuana kadar kullandığımız her şey public'ti.
2. **protected:** Sadece class ve child class'lar tarafından kullanıma açıktır, dışarıya kapalıdır. Variable başına '_' konur, örneğin `_name`.

*NOT: Python'da protected yoktur, '_' başında bulunsa bile kullanıma açıktır, '_' Python için değil yazılımcı için bir uyarıdır!*

3. **private:** Sadece class'ın kendisi tarafından kullanıma açıktır, dışarıdan erişilemez! Child class'lar parent içerisindeki private'lara **erişemez!** Variable başına '__' gelir, örneğin `__get_price()`.

Genel kodlama prensibi olarak bizler mümkün olduğunca attribute'ları gizlemeye, method'ları ise (yardımcı fonksiyon ise protected ya da private tercih edilebilir) mümkün olduğunca dışarıya açmaya çalışırız. Bunun temel sebebi kullanıcıların (ya da kazara yazılımcıların) kontrolsüz bir şekilde kritik verilere dokunması önlemek içindir! Kullanıcı bir asker'in canını sisteme keyfince negatif ya da çok büyük değerler girememiş olur. Ancak method'lar sayesinde ise gene can değerine olan erişimini korur.

![encapuslation](https://media.geeksforgeeks.org/wp-content/uploads/20260522161654499472/encapsulation_in_python.webp)

````python
# Soldier'ı yenileyelim!

class Soldier:
    """
    This is my soldier implementation! 
    It's all up to your imagination!
    """
    soldier_count = 0
    
    def __init__(self, name: str, faction: str, type: SoldierType, hp: float, stamina: float):
        self.name = name
        self.faction = faction
        if not isinstance(type, SoldierType):
            raise TypeError("It should be SoldierType")
        self.type = type
        self._hp = None # Protected versiyonu ekledim!
        self.hp = hp
        if hp > self.calculate_max_hp() or hp < 0:
            raise ValueError("Hp can't be lower than zero or greater than the max")
        self._stamina = None # Protected versiyonu ekledim!
        self.stamina = stamina
        if stamina > self.calculate_max_stamina() or stamina < 0:
            raise ValueError("Stamina can't be lower than zero or greater than the max")
        Soldier.soldier_count += 1

    # Devamını şimdilik görmezden gelelim!
````

#### @property ile Getter ve Setter

Yukarıda protected iki tane attribute ekledik, private da ekleyebilirdik. Ancak bunlara herhangi bir şekilde dışarıdan erişimim yok, sadece class içinde kullanabiliyorum. O zaman ben o veriye nasıl ulaşacağım? Ya da nasıl direkt değiştireceğim / kontrol edeceğim? İşte bunun için **getter** ve **setter** fonksiyonlar mevcut! Her OOP içeren dilde bu konsept var! Peki Python'da nasıl yapıyoruz hemen öğrenelim!

Python bize `@property` adlı bir decorator veriyor. Bunun altında yer alan fonksiyonun isminde yeni bir attribute'u anında bizler için üretiyor. Bu sayede hem olmayan yeni attribute'lar üretebiliyoruz, hem de bir private ya da protected attribute'un public versiyonunu oluşturup gizlenen değeri dönüyoruz.

`@property` bize iki fonksiyon daha hediye ediyor: **setter** ve **deleter**. Bizim odağımız şu anda ilki. `@property` ile bir attribute oluşturduktan sonra `@attribute.setter` formatında bir decorator ile gene attribute isminde bir fonksiyon oluşturuyoruz. Bu oluşan fonksiyon içerisinde bir veri set edilmeden hemen önce onu kontrol etme, hata verme, mümkünse düzeltme gibi işlemleri de yapabailiriz! Bu sayede gizli değerimizi direkt değiştirebiliriz!

````python
# Getter ve Setter ekleyelim!

class Soldier:
    """
    This is my soldier implementation! 
    It's all up to your imagination!
    """
    soldier_count = 0
    
    def __init__(self, name: str, faction: str, type: SoldierType, hp: float, stamina: float):
        self.name = name
        self.faction = faction
        if not isinstance(type, SoldierType):
            raise TypeError("It should be SoldierType")
        self.type = type
        self._hp = None # Ders sırasında yaşadığımız sorunların temel sebebi protected olanı public olandan aşağıya koymamızdı

        print("1. protected set") # Hadi işlem sırasını takip edelim!

        self.hp = hp

        print("3. public set")

        self._stamina = None # Ders sırasında yaşadığımız sorunların temel sebebi protected olanı public olandan aşağıya koymamızdı
        self.stamina = stamina
        Soldier.soldier_count += 1

    # Property ile getter üretebiliriz! 
    # Property bir attribute dönmek zorunda olmayıp altındaki fonksiyon ismine bir attribute gibi davranır!
    @property
    def hp(self):
        return self._hp

    @property
    def stamina(self):
        return self._stamina

    # Yeni bir attribute üretebiliriz!
    @property
    def max_hp(self):
        match self.type:
            case SoldierType.MAGE: return 175
            case SoldierType.CAVALRY: return 250
            case SoldierType.ARCHER: return 125
            case SoldierType.SWORDSMAN: return 200
            case _: return 0  

    @property
    def max_stamina(self):
        match self.type:
            case SoldierType.MAGE: return 300
            case SoldierType.CAVALRY: return 200
            case SoldierType.ARCHER: return 200
            case SoldierType.SWORDSMAN: return 150
            case _: return 0          

    # Property ile yaptığımız yeni attribute için attribute.setter ile set mekanizması kurabiliriz!
    @hp.setter
    def hp(self, value): # Değer almamız lazım kullanıcıdan
        print("2. setter called")
        if value > self.max_hp or value <= 0:
            raise ValueError("Max hp has been exceeded, not accepted")
        self._hp = value

    @stamina.setter
    def stamina(self, value): # Değer almamız lazım kullanıcıdan
        if value > self.max_stamina:
            raise ValueError("Max stamina has been exceeded, not accepted")
        self._stamina = value

# Oluşturulurken sıra nasıl oluyor izleyelim!
my_soldier0 = Soldier("Ahmet", "Khuzait", SoldierType.CAVALRY, 20.0, 150.0)
print("----")

# Getter Kullanalım!
print(my_soldier0.hp)
print("----")

# Setter tekrar kullanalım!
my_soldier0.hp = 30.0
````

### Part 2.3: Polymorphism

Polymorphism de bir method, fonksiyon ya da işlemin üzerinde çalıştığı object ya da veri'ye göre farklı çalışmasına denir. Polymorphism'in en basit örneği çarpma/toplama işlemleridir! Bu işlemler iki tarafındaki veri türlerine göre oldukça farklı davranışlara sahiptirler. Basitçe listelemek gerekirse (toplama üzerinden):

Toplama İşlemleri
1. int + int -> int
2. int + float -> float
3. float + int -> float
4. float + float -> float
5. bool + int -> int
6. bool + bool -> int
7. int + bool -> int
8. bool + float -> float
9. float + bool -> float

Pozitifini Alma
10. +int -> int
11. +float -> float

Birleştirme
12. str + str -> str
13. list + list -> list
14. tuple + tuple -> tuple

Burada sadece ilk aklıma gelenleri sıralayabildim, ancak anlaşıldığı üzere veri tiplerine göre birçok işlem tanımlanabilir. Bu da bir çok forma bürünmek yani polymorphism olur. Polymorphism'in iki türü vardır:

1. **compile-time:** Method overloading buraya girer. Yani yukarıdaki örnekte de olduğu gibi, bir method ya da fonksiyonun birden fazla veri türü üzerinde tanımlı olmasıdır.
2. **run-time:** Method overriding de buraya girer. Bunu ise inheritance kısmında gördük, yaptığımız override'lar sayesinde method'lar yaptığı işleme onu çağıran object'in türüne göre karar verir.

### Part 2.4: Abstraction

Abstraction, arka planda dönen o karmaşık kod yığınını, büyük mantıksal işlemleri kullanıcıdan saklayıp sadece en gerekli kısımları göstermek anlamına gelir. Normalde zaten bunu class'lar ile yapıyoruz. Ancak bunu daha gelişmiş düzeyde yapmanın bir yolu var: Abstract class'lar. Abstract class'lar diğer class'ların kendisini inherit edip şema olarak kullandığı class'lardır. Bu class'lar içerisinde çok az method hazır bulunur ya da sadece method'un adı yer alıp iç kısmı `pass` ile boş bırakılmıştır. Abstract class'lar direkt tek başlarına **asla** kullanılmazlar.
```python
from abc import ABC, abstractmethod

# Abstract Base Class
class Vehicle(ABC):

    # Abstract method
    @abstractmethod
    def move(self):
        pass

    # Concrete method
    def fuel_status(self):
        return "Checking fuel level..."

# Concrete Subclasses
class Car(Vehicle):
    def move(self):
        return "The car drives on the road."

class Boat(Vehicle):
    def move(self):
        return "The boat sails on the water."

my_car = Car()
print(my_car.move()) # Abstract method'un kendi versiyonu
print(my_car.fuel_status()) # Abstract class'a ait concrete method

vehicle = Vehicle() # TypeError verir, çünkü abstract class'tan object üretilemez!
```
Yukarıda örneği yer alan abstract class'ların 4 temel parçası vardır (bu örnekte 3 tane var):
1. **Abstract Base Class (ABC):** Class'ın kendisidir, direkt bir object üretemez ancak inherit edilebilir.
2. **Abstract Method:** `@abstractmethod` ile oluşturduğumuz method türüdür, `pass` kelimesi ile bırakılır, içeriği yazılmaz. Inherit eden child class'ta tanımlanmak zorundadır.
3. **Concrete Method:** Standard yazdığımız method'lardır. Bunlar tamamen yazılmış olabilir, kısmen yazılmış olabilir. Child class tanımlamak zorunda değildir, direkt kullanabilir.
4. **Abstract Property:** Abstract Method gibi child class tarafından tam olarak tanımlanması zorunlu olan property/attribute'lara verilen isimdir.
