### Part 1.1: Exception'lar ve Kullanımı
Bugüne kadar hep Python bize hata verdi, yapamazsın dedi edemezsin dedi. Artık sıra bizde! Artık biz de kullanıcıya hata vereceğiz! Python'ın şimdiye kadar bize verdiği hata (error)'ların genel ismi *exception*. Bizler de bunları kullanıp kullanıcı bir hata yaptığında onu uyarabiliriz! Bu *exception*'ları `raise` kelimesi ile beraber çağırmaktayız. En çok karşılaştığımız ve OOP'de kullandığımız iki adet *exception* bulunmakta:

* **`ValueError()`:** Gelen veri/değer istediğimiz değerlerin dışında kalıyorsa, örneğin bir can değeri için kullanıcı negatif değer girmişse bu *exception*'ı kullanabiliriz.
* **`TypeError()`:** Gelen veri/değer istediğimiz veri tipi ile uyuşmuyorsa, örneğin float istenen yerde kullanıcı string girmişse bu *exception*'ı kullanabiliriz.

ValueError'lar için sıradan karşılaştırmalar iş görmekte, ancak TypeError konusunda şu ana kadar öğrendiklerimizle zorlanabiliriz! Bunun için çok güzel bir yardımcı fonksiyon var: `isinstance(obj, cls)` bu fonksiyon, eğer verilen object o class'ın bir object'i ise *True*, değilse *False* döner. Derste bahsetmiştik, `int`, `str`, `list` gibi veri tipleri de Python'da bir class, dolayısıyla onlar için de kullanabiliriz!

````python
# Eski örneği geliştirelim,
# Şuanki odak Faction'da

class Soldier:
    """
    This is my soldier implementation! 
    It's all up to your imagination!
    """
    soldier_count = 0
    
    def __init__(self, name: str, faction: str, type: str, hp: float, stamina: float):
        self.name = name
        self.faction = faction
        self.type = type
        self.hp = hp
        self.stamina = stamina
        Soldier.soldier_count += 1

    def __str__(self):
        return f"{self.faction}'den {self.name} adlı asker"

    def __repr__(self):
        return f"Soldier(name={self.name}, faction={self.faction}, type={self.type})"

class Faction:
    """
    This is my faction implementation! 
    It's all up to your imagination!
    """
    all_factions = []
    
    @classmethod
    def list_factions(cls):
        return cls.all_factions

    def __init__(self, name, lands, troops, ruler, power ,enemies = [], friends = []):
        self.name = name
        self.lands = lands
        self.troops = troops
        self.ruler = ruler

        if power < 0:
            raise ValueError(f"Power can't be negative, {power} given.")
        self.power = power
        self.enemies = enemies
        self.friends = friends
        Faction.all_factions.append(self)


    def can_hire(self, soldier: Soldier):
        # Türünü kontrol ediyorum!
        if not isinstance(soldier, Soldier):
            # raise ile çağırdım
            raise TypeError("Invalid object, Soldier expected, {} given".format(type(soldier).__name__)) # Parantez içinde error için mesaj verebiliriz!
            # __name__ dunder'ı direkt class ismini verir!
        if soldier.faction != self.name:
            return False
        
        if soldier.hp >= 100.0 and soldier.stamina >= 200.0:
            return True
        elif soldier.hp >= 250.0 and soldier.stamina >= 100.0:
            return True
        else:
            return False

    def __str__(self):
        return f"{self.ruler}'ın Büyük {self.name} Krallığı"

my_new_faction = Faction("Aserai", [], [], "Erenalp", 1)

# my_new_faction.can_hire(10) # 10 bir solider değil, hata verecek!

# my_other_new_faction = Faction("Aserai", [], [], "Erenalp", -3) # -3 kabul etmiyorum, hata verecek!
````

#### Gelişmiş Exception Kullanımı

Bazen Exception'lar ile çalışırken özel işlemler yapmamız gerekbilir, kullanıcının yüzüne bağırmak yerine gelen hata türüne göre eylem almamız gerekebilir (lütfen kimsenin yüzüne bağırmayın please). Bunun için Python'da tıpkı `if-else` gibi `try-except` zincirleri bulunmakta. Hadi beraber inceleyelim:
```python
def foo(a,b,c):
    flag = 0 # Hata yaşadım mı diye tespit etmek için flag koydum!
    try:
        res = (a+b)/c # c 0 ise ZeroDivisionError verilir! 
        # Bölme işleminden dolayı bir tanesi bile float ya da int değilse de TypeError verilir!

        if c < 0:
            raise ValueError("c can't be a negative number") # Bu kuralı da ben koydum!

    except ZeroDivisionError: # Hata oluştuğu gibi yakaladım!
        print("c can't be zero")
        flag = 1
    except TypeError:
        print("all variables should be either int or float")
        flag = 1
    except ValueError as e: # as e ile orijinal mesajı yakalayabiliriz!
        print("ValueError catched:",e)
        flag = 1

    else: # Hiç exception yoksa çalışır.
        return res

    finally: # Bu her zaman çalışır, ne olursa olsun!
        if flag:
            print("An exception occured, exiting without return")
        else:
            print("No errors occured, the result:",res)
```
Yukarıdaki yapıyı beraber inceleyelim. Yukarıdaki fonksiyonda `try`, `except`, `else` ve `finally` blokları yer alsa da sadece ilk ikisi zorunludur, sadece `except` bloğu birden fazla olabilir. `try` bloğu içerisinde yaptığımız işlemde Python tarafından ya da bizim tarafımızdan çeşitli hatalar türeyebilir. Bunları ise tek tek ya da birlikte (`except (ZeroDivisionError, TypeError):` gibi) `except` bloklarında yakalayıp özel işlemlerimizi halledebiliriz. `as e` yapısı ile yakaladığımız *exception*'ın mesajını direkt olarak elde edebiliriz! Böylece özellikle kendimiz oluşturduğumuz ya da çağırdığımız *exception*'lar için tekrar bir şey yazmaya gerek kalmaz. Veyahut genel bir error yakalamak için:
```python
try:
    result = 10 / 0
except Exception as e:
    print(f"Error type: {type(e).__name__}")
    print(f"Error message: {e}")
```
örneğindeki gibi herhangi bir *exception*'ı `e` olarak yakalayıp onun önce türünü (ValueError, ZeroDivisionError ...) sonra da mesajını verebiliriz! `else` bloğu ise hiçbir exception yaşanmadıysa çalışır, buraya da dolayısı ile return değerimizi yerleştirdik. `finally` bloğu ise her şart altında çalışması garantidir. Hatta, yukarısındaki bloklarda return varsa Python hemen `finally`'e atlar, onu tamamlar ve sonra return işlemini gerçekleştirir.

*ÖNEMLİ NOT: `try`, `except` ve `else` bloklarında return bulunabilir, ancak `finally` bloğuna asla return koymamalısınız! `finally`'de return ifadesi hata verme işlemini iptal eder, tüm exception'lar kaybolur.*

*dipnot: Python 3.14 ile (çok şükür sonunda akıl etmişler) `finally` içinde yer alan return'ler hata vermekte.*

````python
# Örneği deneyelim!

def foo(a,b,c):
    flag = 0 # Hata yaşadım mı diye tespit etmek için flag koydum!
    try:
        res = (a+b)/c # c 0 ise ZeroDivisionError verilir! 
        # Bölme işleminden dolayı bir tanesi bile float ya da int değilse de TypeError verilir!

        if c < 0:
            raise ValueError("c can't be a negative number") # Bu kuralı da ben koydum!

    except ZeroDivisionError: # Hata oluştuğu gibi yakaladım!
        print("c can't be zero")
        flag = 1
    except TypeError:
        print("all variables should be either int or float")
        flag = 1
    except ValueError as e: # as e ile orijinal mesajı yakalayabiliriz!
        print("ValueError catched:",e)
        flag = 1

    else: # Hiç exception yoksa çalışır.
        return res

    finally: # Bu her zaman çalışır, ne olursa olsun!
        if flag:
            print("An exception occured, exiting without return")
        else:
            print("No errors occured, the result:",res)

print("--Normal--")
foo(3,4,5.7)

print("\n--0'a Bölme--")
foo(3,4,0)

print("\n--Negatif C Kuralı--")
foo(3,4,-7)

print("\n--String--")
foo("a",4,2)
````

### Part 1.2: Decorator'lar

Ders sırasında `@classmethod` ve `@staticmethod` yapılarını görmüştük. Bunlar peki kim ki bunların başında '@' var? olay ne? Biz bu tür başında '@' olan ve fonksiyonları ya da method'ları modifiye eden fonksiyonlara (evet onlar da aslında fonksiyon) decorator diyoruz! Decorator'lar genel anlamda şu yapıda tanımlanırlar:
```python
def decorator(func):
    def wrapper():
        func()
    return wrapper

@decorator
def foo():
    pass
```
Örnekte de olduğu gibi decorator fonksiyonumuz argüman olarak bir fonksiyon alır. İçinde bir fonksiyon daha tanımlanır, (normalde içerideki fonksiyonu kullanabilmemiz için onu çağırmamız lazım, ancak decorator sırasında bu yok!) ve bu fonksiyon return'lenir. Return'lenen *wrapper* fonksiyon içinde asıl fonksiyonumuz ile ilgili çeşitli işlemler yapılabilir. En son da decorator fonksiyonumuz başında '@' ile beraber kullanılır, altında yer alan fonksiyon argüman olarak kullanılır.

````python
# Örneklemeden anlayamayız!

# Decorator fonkisyonumu tanımladım
def readable(fun):
    # Ve wrapper fonksiyonum!
    def merge():
        # İstediğim tüm işlemler buarada
        return "Do you know the story of: " + fun()
    return merge
    # NOT: Decorator içerisinde sadece return wrapper olmalı
    # Ek bir işlem olmamalı!

# Kolayca modifiye ettim!
@readable
def foo():
    return "FU"

@readable
def bar():
    return "BAR"

print(foo())
print(bar())
````

#### Argüman Alan Fonksiyonlar İçin Decorator'lar

Peki ya fonksiyon argüman alıyorsa? Onun için çözüm çok basit! `*args` ile `**kwargs`'ı geçen derslerde görmüştük! İşte tam da burada onları kullanmamız lazım! Aşağıdaki format ile argüman alan tüm fonksiyonlar için decorator yazabiliriz!
```python
def decorator(func):
    def wrapper(*args, **kwargs):
        func(*args, **kwargs)
    return wrapper

@decorator
def foo(x, y):
    pass
```
Burada Python fonksiyona normalde gelen tüm argümanları rahatlıkla kullanır!

#### Method'lar İçin Decorator'lar

Şu ana kadar yazdığımız decorator'lar maalesef method'lar için yeterli değil! Çünkü bir şey eksik... Evet `self` eksik! Onu da eklersek artık method'lar için bir decorator yazmış oluruz!
```python
def decorator(func):
    def wrapper(self, *args, **kwargs):
        func(self, *args, **kwargs)
    return wrapper

class Bar:
    @decorator
    def foo(self, x, y):
        pass
```
Böylelikle Python method'lar ile olan işlemleri de halletmiş oluyor!

![aynı ben](https://static.boredpanda.com/blog/wp-content/uploads/2025/11/6911a547711f3_35ezb6d118lf1-jpeg__880.jpg)

````python
# Taylor estimation'ı bambaşka bir hale getirelim!
from math import *

# Öyle bir decorator yazalım ki, cos, sin ve e^x için estimation yapabilsin!
def taylor_general(fun):
    def taylor(*args, **kwargs):
        # Fonksiyonun output değerinden türünü anlayalım!
        sum = 0
        if e == fun(1): # e^x ise
            for i in range(11): # 10. dereceden estimation
                num_pow = pow(args[0],i) # args[0] fonksiyona verilen sayı!
                fact = factorial(i)
                sum += (num_pow/fact)

        elif fun(30) == cos(radians(30)): # cos ise
            deg = args[0] % 360
            rad = radians(deg) # dereceyi radyan yapıyoruz!
            for i in range(4):
                neg_pow = pow(-1,i)
                num_pow = pow(rad,(2*i))
                fact = factorial(2*i)
                sum += ((neg_pow * num_pow) / fact)

        elif fun(30) == sin(radians(30)): # sin ise
            deg = args[0] % 360
            rad = radians(deg) # dereceyi radyan yapıyoruz!
            for i in range(4):
                neg_pow = pow(-1,i)
                num_pow = pow(rad,(2*i)+1)
                fact = factorial((2*i)+1)
                sum += (neg_pow * num_pow) / fact

        return sum
    return taylor
# Ben niye böyle bişi yaptım? Keşke bilsem, yazık kafama

# -- E^x --
print("original",exp(3))

@taylor_general
def my_exp(num):
    return exp(num)

print("estimation",my_exp(3))

# -- COS --
print("original",cos(radians(30)))

@taylor_general
def my_cos(num):
    return cos(radians(num))

print("estimation",my_cos(30))

# -- SIN --
print("original",sin(radians(30)))

@taylor_general
def my_sin(num):
    return sin(radians(num))

print("estimation",my_sin(30))


# İnanın niye bunu yaptım bilmiyorum
````

### Part 1.3: Enumerator'lar

Enumerator'lar ya da kısaca enum, Python ve diğer dillerde *constant*'ları bir arada tutmak için kullanılır. Bu sayede eğer bir yerden ya da kullanıcıdan gelen işlemleri veyahut sistem içinde çalışan mekanizmaları sabit kılmak istiyorsak, yazım hatalarına karşı korunaklı kılmak istiyorsak bunu rahatlıkla yapabiliriz. Enum kullanabilmek için Python'ın built-in enum library'sinden import etmemiz gerekir.
```python
from enum import Enum

class Society(Enum):
    CS = "CS"
    COMSOC = "ComSoc"
    RAS = "RAS"
    PES = "PES"
    AESS = "AESS"
```
Yukarıdaki örnekte de olduğu gibi enum sayesinde sahip olduğumuz *constant*'ları (Büyük harflerle yazılmış olan isimler) gruplamış olduk. Artık kullanıcının girdilerini sistem daha doğru bir şekilde tespit edebilir, ve iç mekanizmaları `Society.RAS` gibi şekilde enum kullanıp kararlılık sağlarız ve yazım hatalarından korunuruz! Kıyaslamalarımızı direkt `obj.attr == Society.AESS` şeklinde yapabiliriz! İsimleri görmek istiyorsak `Society.CS.name` yazıp *CS*, karşılık geldikleri değerleri merak ediyorsak `Society.COMSOC.value` yazıp *ComSoc* çıktısını alabiliriz. Unutmayalım, value bir string olmayabilir! `PI = 3.14` yazmak da mümkün!
